```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
import shutil
from pathlib import Path
```

```python
date = '260703'
session_home = '4'
session_etho = '1'
path_session = Path('/srv/MarmOT/VideoTracking/Videos') / date
#home3d = os.path.join('/scratch','VideoTracking','Videos',f'TestHomeWithEtho{session}','Output','videos_labeled_3d','optimized_composite.mp4')
#etho3d = os.path.join('/scratch','VideoTracking','Videos',f'TestEthoWithHome{session}','Output','videos_labeled_3d','optimized_composite.mp4')
home2d = path_session / 'Home' / 'Input_preprocessed' / 'output4.mp4'
etho2d = path_session / 'Etho' / 'Input_preprocessed' / 'output1.mp4'
vidcap_home = cv2.VideoCapture(home2d)
vidcap_etho = cv2.VideoCapture(etho2d)
#vidcap_home3d = cv2.VideoCapture(home3d)
#vidcap_etho3d = cv2.VideoCapture(etho3d)
start_frame_etho = 1369
start_frame_home = 1520
stop_frame_home = 39718
stop_frame_etho = 40422
jitter_start = 10
duration = 20
step = 1
```

```python
vidcap_home.set(cv2.CAP_PROP_POS_FRAMES, start_frame_home - jitter_start * step)
vidcap_etho.set(cv2.CAP_PROP_POS_FRAMES, start_frame_etho - jitter_start * step)
shutil.rmtree('tmp2') 
os.mkdir('tmp2')

for i in range(-jitter_start*step, step*(duration - jitter_start)):
    print(i)
    success_home, frame_home = vidcap_home.read()
    success_etho, frame_etho = vidcap_etho.read()
    if i % step == 0:
        fig, axs = plt.subplots(1,2,figsize=(40,20))
        axs[0].imshow(frame_etho[:,:,::-1])
        axs[0].axis('off')
        axs[0].set_title(f'Etho frame {i + start_frame_etho}')
        axs[1].imshow(frame_home[:,:,::-1])
        axs[1].axis('off')
        axs[1].set_title(f'Home frame {i + start_frame_home}')
        fig.savefig(f'tmp2/{i}.png')
        plt.cla()
        plt.close(fig)

```

```python
vidcap_home.set(cv2.CAP_PROP_POS_FRAMES, stop_frame_home - jitter_start)
vidcap_etho.set(cv2.CAP_PROP_POS_FRAMES, stop_frame_etho - jitter_start)
for i in range(-jitter_start, 25 - jitter_start):
    success_home, frame_home = vidcap_home.read()
    success_etho, frame_etho = vidcap_etho.read()
    fig, axs = plt.subplots(1,2,figsize=(40,20))
    axs[0].imshow(frame_etho[:,:,::-1])
    axs[0].axis('off')
    axs[0].set_title(f'Etho frame {i + stop_frame_etho}')
    axs[1].imshow(frame_home[:,:,::-1])
    axs[1].axis('off')
    axs[1].set_title(f'Home frame {i + stop_frame_home}')
    fig.show()

```
